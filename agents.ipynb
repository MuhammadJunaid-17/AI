```python
from google import genai

client = genai.Client(api_key="")
```

```python
MODEL = "models/gemini-3.6-flash"
```

```python
def travel_agent(user_input):

    prompt = f"""
You are a professional travel agent for Pakistan tourism.

Provide a well-structured travel plan with clear headings:

1. Destination Overview

Make response detailed but easy for students.

User request: {user_input}

Travel Plan:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text



print("Gemini Travel Agent Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye ")
        break

    try:
        reply = travel_agent(user)
        print("\nAgent:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```

```python
def global_stylist(user_input):

    prompt = f"""
You are a professional global wheather agent for user.

Provide a well-structured plan with clear headings:

1. The wheather details like temperature.
2. And give clothes update according to wheather like warm clothes in winter.
3. make it briefly described.
Make response detailed but easy for students.

User request: {user_input}

Global stylist:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text



print("Welcome to global stylist Agent (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye ")
        break

    try:
        reply = global_stylist(user)
        print("\nAgent:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```

```python
# Student Data
student_data = {
    "ali": {
        "math": 85,
        "english": 78,
        "physics": 90
    },
    "ahmed": {
        "math": 72,
        "english": 88,
        "physics": 80
    },
    "sara": {
        "math": 95,
        "english": 91,
        "physics": 87
    }
}


# Gemini fallback function
def student_agent(user_input):

    prompt = f"""
You are a student marks assistant.

Generate a reasonable example mark between 0 and 100.

Return ONLY the number of marks.

User request:
{user_input}
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text.strip()


# Main Program
print("Welcome to Student Marks Agent (type 'exit' to stop)\n")

while True:

    name = input("Enter student name: ").strip().lower()

    if name == "exit":
        print("Goodbye!")
        break

    subject = input("Enter subject: ").strip().lower()

    if subject == "exit":
        print("Goodbye!")
        break

    try:

        if name in student_data:

            if subject in student_data[name]:

                # Marks found in dictionary
                marks = student_data[name][subject]

            else:

                # Subject not found, ask Gemini
                user_input = f"""
Student: {name}
Subject: {subject}
"""

                marks = student_agent(user_input)

        else:

            # Student not found, ask Gemini
            user_input = f"""
Student: {name}
Subject: {subject}
"""

            marks = student_agent(user_input)

        display_name = name.title()
        display_subject = subject.title()

        print(f"\nAgent: {display_name} got {marks} marks in {display_subject}.\n")

    except Exception as e:

        print("Error:", e)
```
