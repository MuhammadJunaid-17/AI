```python
print("--- Welcome to Rocky Travel Agency ---\n")

# Step 1: Menu dikhana
print("Available Holiday Packages:")
print("1. Paris  - $1200 per person")
print("2. Tokyo  - $1500 per person")
print("3. London - $1000 per person")
print("4. Dubai  - $800 per person\n")

# Step 2: User se destination poochna
choice = int(input("Enter the number of your destination (1-4): "))

# Switch case ki jagah if-elif use kar rahe hain
if choice == 1:
    price_per_ticket = 1200
    print("\nDestination Selected: Paris")

elif choice == 2:
    price_per_ticket = 1500
    print("\nDestination Selected: Tokyo")

elif choice == 3:
    price_per_ticket = 1000
    print("\nDestination Selected: London")

elif choice == 4:
    price_per_ticket = 800
    print("\nDestination Selected: Dubai")

else:
    print("Invalid choice! Please restart the program.")
    exit()

# Step 3: Tickets ki tadad poochna
tickets = int(input("How many tickets would you like to book? "))

if tickets <= 0:
    print("Invalid number of tickets!")
    exit()

# Step 4: Total bill calculate karna
total_cost = price_per_ticket * tickets

print("\n--- Booking Summary ---")
print("Tickets booked:", tickets)
print("Total Amount Due: $", total_cost)
print("Thank you for booking with us! Have a safe trip.")
```
